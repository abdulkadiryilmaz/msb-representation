# Stage 1A Latent Inspection

Bu notebook, Stage 1A checkpoint latent artefaktlarını görsel olarak incelemek için hazırlandı.

Amaçlar:

- PCA ve UMAP scatter'larını görmek
- label / symbol / bucket katmanlarını aynı uzayda incelemek
- misclassified örnekleri seçip nearest-neighbor komşularını okumak
- seçili örnek için short-window fiyat davranışını görmek

Kullanım:

- `CHECKPOINT_NAME` ile checkpoint seçin: `ce_only`, `ce_supcon_v1`, `ce_supcon_long_v1`
- `SPLIT` ile `val` / `test` arasında geçin
- Aynı notebook ile CE-only ve SupCon varyantlarını yan yana manuel okuyun


```python
from __future__ import annotations

from pathlib import Path
import json
import sys

REPO_ROOT = Path.cwd()
if not (REPO_ROOT / "src").exists() and (REPO_ROOT.parent / "src").exists():
    REPO_ROOT = REPO_ROOT.parent

sys.path.insert(0, str(REPO_ROOT / "src"))
sys.path.insert(0, str(REPO_ROOT / "scripts"))
sys.path.insert(0, str(REPO_ROOT))

import numpy as np
import polars as pl
import plotly.graph_objects as go
from plotly.subplots import make_subplots

from msb_repr.stage1a.analysis import load_latent_export
from msb_repr.stage1a.dataset import Stage1ADualWindowDataset
from msb_repr.stage1a.config import Stage1ADatasetSpec

try:
    from analyze_stage1a_latents import _assign_domain_bucket, _build_bucket_thresholds
except ModuleNotFoundError:
    from scripts.analyze_stage1a_latents import _assign_domain_bucket, _build_bucket_thresholds
```

```python
DATASET_ROOT = REPO_ROOT / "data/stage1a/binance/15m"

# "ce_only", "ce_supcon_v1", "ce_supcon_long_v1"
CHECKPOINT_NAME = "ce_supcon_long_v1"
CHECKPOINT_DIR = DATASET_ROOT / f"checkpoints/{CHECKPOINT_NAME}"

# 'val' veya 'test'
SPLIT = "test"

ANALYSIS_DIR = CHECKPOINT_DIR / "analysis" / f"{SPLIT}_latents"
LATENT_PATH = CHECKPOINT_DIR / "analysis" / f"{SPLIT}_latents.npz"

assert LATENT_PATH.exists(), f"Latent export bulunamadı: {LATENT_PATH}"
assert ANALYSIS_DIR.exists(), f"Analysis dir bulunamadı: {ANALYSIS_DIR}"

print("Using checkpoint:", CHECKPOINT_NAME)
print("Using split:", SPLIT)
print("Latents:", LATENT_PATH)
print("Analysis dir:", ANALYSIS_DIR)
```

```python
arrays, export_metadata = load_latent_export(LATENT_PATH)
spec = Stage1ADatasetSpec.load(DATASET_ROOT / "spec.json")
dataset = Stage1ADualWindowDataset.from_root(DATASET_ROOT, split=SPLIT)
thresholds = _build_bucket_thresholds(arrays)
min_recent_break_bars = int(export_metadata["label_config"]["min_recent_break_bars"])

pca_df = pl.read_csv(ANALYSIS_DIR / "pca_projection.csv")
umap_path = ANALYSIS_DIR / "umap_projection.csv"
umap_df = pl.read_csv(umap_path) if umap_path.exists() else None
nn_df = pl.read_csv(ANALYSIS_DIR / "nearest_neighbors.csv")
summary = json.loads((ANALYSIS_DIR / "summary.json").read_text())

print(summary["accuracy"], summary["mean_confidence"], summary["mean_neighbor_label_agreement"])
```

```python
buckets = []
bucket_reasons = []
for idx in range(len(arrays["labels"])):
    bucket, reason = _assign_domain_bucket(arrays, idx, min_recent_break_bars, thresholds)
    buckets.append(bucket)
    bucket_reasons.append(reason)

meta_df = pl.DataFrame(
    {
        "index": np.arange(len(arrays["labels"])),
        "symbol": arrays["symbols"],
        "timestamp": arrays["timestamps"],
        "label": arrays["labels"],
        "pred": arrays["preds"],
        "label_name": [str(x) for x in arrays["label_names"][arrays["labels"]]],
        "pred_name": [str(x) for x in arrays["label_names"][arrays["preds"]]],
        "confidence": arrays["probs"].max(axis=1),
        "misclassified": arrays["labels"] != arrays["preds"],
        "bucket": buckets,
        "bucket_reason": bucket_reasons,
    }
)

pca_view = pca_df.join(meta_df, on="index", how="left")
umap_view = umap_df.join(meta_df, on="index", how="left") if umap_df is not None else None

display(meta_df.head())
```

## PCA / UMAP Scatter Helpers

```python
COLOR_MAP = {
    "bullish": "#d1495b",
    "bearish": "#00798c",
    "intact": "#edae49",
}

def scatter_projection(df: pl.DataFrame, title: str, color_col: str = "label_name", only_misclassified: bool = False):
    plot_df = df.filter(pl.col("misclassified")) if only_misclassified else df
    fig = go.Figure()
    for key in plot_df[color_col].unique().to_list():
        sub = plot_df.filter(pl.col(color_col) == key)
        marker_color = COLOR_MAP.get(key, None)
        fig.add_trace(
            go.Scattergl(
                x=sub["x"].to_list(),
                y=sub["y"].to_list(),
                mode="markers",
                name=str(key),
                marker={"size": 6, "opacity": 0.7, **({"color": marker_color} if marker_color else {})},
                customdata=np.stack(
                    [
                        sub["index"].to_numpy(),
                        sub["symbol"].to_numpy(),
                        sub["timestamp"].to_numpy(),
                        sub["label_name"].to_numpy(),
                        sub["pred_name"].to_numpy(),
                        sub["bucket"].to_numpy(),
                        sub["confidence"].to_numpy(),
                    ],
                    axis=1,
                ),
                hovertemplate=(
                    "idx=%{customdata[0]}<br>"
                    "symbol=%{customdata[1]}<br>"
                    "timestamp=%{customdata[2]}<br>"
                    "label=%{customdata[3]} pred=%{customdata[4]}<br>"
                    "bucket=%{customdata[5]}<br>"
                    "confidence=%{customdata[6]:.4f}<extra></extra>"
                ),
            )
        )
    fig.update_layout(title=title, height=700, width=1000)
    fig.show()
```

```python
scatter_projection(pca_view, f"{SPLIT.upper()} PCA by label", color_col="label_name")
```

```python
scatter_projection(pca_view, f"{SPLIT.upper()} PCA by symbol", color_col="symbol")
```

```python
scatter_projection(pca_view, f"{SPLIT.upper()} PCA by bucket", color_col="bucket")
```

```python
scatter_projection(pca_view, f"{SPLIT.upper()} PCA misclassified only", color_col="bucket", only_misclassified=True)
```

## Bucket Summary

```python
bucket_summary = (
    meta_df.group_by(["bucket", "label_name"])
    .agg(
        pl.len().alias("count"),
        pl.col("misclassified").mean().alias("misclassified_rate"),
        pl.col("confidence").mean().alias("avg_confidence"),
    )
    .sort(["count"], descending=True)
)
display(bucket_summary)
```

## Nearest-Neighbor Inspection

```python
from datetime import datetime, timezone

def _format_epoch_seconds(value: int) -> str:
    return datetime.fromtimestamp(int(value), tz=timezone.utc).strftime("%Y-%m-%d %H:%M:%S UTC")

def with_readable_timestamps(df: pl.DataFrame) -> pl.DataFrame:
    out = df
    for col in df.columns:
        if "timestamp" in col:
            out = out.with_columns(
                pl.col(col).map_elements(_format_epoch_seconds, return_dtype=pl.String).alias(f"{col}_dt")
            )
    return out

misclassified_examples = meta_df.filter(pl.col("misclassified")).sort("confidence")
display(
    with_readable_timestamps(misclassified_examples)
    .select(["index", "symbol", "timestamp", "timestamp_dt", "label_name", "pred_name", "bucket", "confidence"])
    .head(25)
)
```

```python
# Buraya incelemek istediğiniz örneğin index'ini yazın.
SELECTED_INDEX = 8740

selected_meta = with_readable_timestamps(meta_df.filter(pl.col("index") == SELECTED_INDEX))
display(selected_meta)

selected_nn = with_readable_timestamps(nn_df.filter(pl.col("index") == SELECTED_INDEX))
display(selected_nn)
```

```python
feature_index = {name: idx for idx, name in enumerate(spec.short_feature_columns)}
open_idx = feature_index["open"]
high_idx = feature_index["high"]
low_idx = feature_index["low"]
close_idx = feature_index["close"]

def short_window_frame(index: int) -> pl.DataFrame:
    sw = dataset.short_windows[index]
    return pl.DataFrame(
        {
            "bar": np.arange(sw.shape[1]),
            "open": sw[open_idx],
            "high": sw[high_idx],
            "low": sw[low_idx],
            "close": sw[close_idx],
        }
    )

def plot_short_window(index: int, title_prefix: str = "sample"):
    df = short_window_frame(index)
    fig = go.Figure(
        data=[
            go.Candlestick(
                x=df["bar"].to_list(),
                open=df["open"].to_list(),
                high=df["high"].to_list(),
                low=df["low"].to_list(),
                close=df["close"].to_list(),
                name=str(index),
            )
        ]
    )
    fig.add_vline(x=spec.labels.structure_bars - 0.5, line_dash="dash", line_color="gray")
    fig.update_layout(
        title=f"{title_prefix} idx={index}",
        xaxis_title="bar",
        yaxis_title="price",
        xaxis_rangeslider_visible=False,
        height=500,
        width=1100,
    )
    fig.show()
```

```python
plot_short_window(SELECTED_INDEX, title_prefix="selected")
```

```python
row = nn_df.filter(pl.col("index") == SELECTED_INDEX).to_dicts()[0]
neighbor_indices = [int(row[f"nn{i}_index"]) for i in range(1, 4)]
neighbor_indices
```

```python
fig = make_subplots(rows=len(neighbor_indices) + 1, cols=1, shared_xaxes=False, vertical_spacing=0.03)

all_indices = [SELECTED_INDEX] + neighbor_indices
for row_no, idx in enumerate(all_indices, start=1):
    df = short_window_frame(idx)
    meta_row = meta_df.filter(pl.col("index") == idx).to_dicts()[0]
    fig.add_trace(
        go.Candlestick(
            x=df["bar"].to_list(),
            open=df["open"].to_list(),
            high=df["high"].to_list(),
            low=df["low"].to_list(),
            close=df["close"].to_list(),
            name=str(idx),
            showlegend=False,
        ),
        row=row_no,
        col=1,
    )
    fig.add_vline(x=spec.labels.structure_bars - 0.5, line_dash="dash", line_color="gray", row=row_no, col=1)
    fig.update_yaxes(title_text=f"{idx} | {meta_row['label_name']}->{meta_row['pred_name']}", row=row_no, col=1)

fig.update_layout(
    title=f"selected sample and nearest neighbors | idx={SELECTED_INDEX}",
    height=300 * (len(all_indices)),
    width=1100,
    xaxis_rangeslider_visible=False,
)
fig.show()
```
